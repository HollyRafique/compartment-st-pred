# 8 - Merge CV Results

Utility functionality to merge the results from all ML models and foundation model feature sets

```python
import pandas as pd
import os
import glob
```

```python
def merge_csvs(output_path,feat_extr, segment = 'PanCK+', ml_models=['elasticnet','gradboost','randomforest','SVM']):
    


    allgenescoreslist = []  
    for geneset in ['Panel','HVG','SVG']:
        print(geneset)
        print(os.path.join(output_path,feat_extr,geneset, "*/"))
        folders = glob.glob(os.path.join(output_path,feat_extr,geneset, "*/"))
        print(len(folders))
        for folder in folders:
            rungenescorelist = []
            folder_path = os.path.join(output_path,feat_extr,folder)
            #print(folder_path)
            for model in ml_models:
            #for model in ['SVM']:    
                df = pd.read_csv(os.path.join(folder_path,f'per_gene_scores-{model}-{segment}.csv'))
                # display(df.head())
                rungenescorelist.append(df)
            #display(df.head(2))
            # per_gene_scores-elasticnet-PanCK+
            genescores = pd.concat(rungenescorelist)
            genescores.insert(0,'GeneSet',geneset)
            allgenescoreslist.append(genescores)

    #print(allgenescoreslist)
    allgenescores = pd.concat(allgenescoreslist)

    allgenescores.insert(1, 'Segment', segment)
    allgenescores.insert(2,'Feature_Extractor', feat_extr)
    allgenescores = allgenescores.rename(columns={"Run": "ML_Model"})

    # do NOT drop duplicates yet as we use this to get our summary results
    # allgenescores = allgenescores.drop_duplicates(subset=['ML_Model','Gene']).reindex()
    print(allgenescores.shape)
    allgenescores.to_csv(os.path.join(output_path,feat_extr,f"allgenescores_{segment[:-1].lower()}_{feat_extr}.csv"),index=False)

    return allgenescores

    
```

# Append Multiple CSVs

```python
segment = 'PanCK+'
segment[:-1].lower()
```

```python

root_path = ".../data"
fms = ['uni','gigapath','virchow','combined']
output_path=os.path.join(root_path,'03-results_cv')
output_path
```

```python

full_scores = []
for feat_extr in fms:
    print(feat_extr)
    scores = merge_csvs(os.path.join(output_path,'PanCK'),feat_extr, segment='PanCK+',ml_models=['SVM'])
    full_scores.append(scores)

panck_scores = pd.concat(full_scores)
panck_scores.to_csv(os.path.join(output_path,f"allgenescores_panck.csv"),index=False)
```

merge the feat extractor ones into one

```python

full_scores = []
for feat_extr in fms:
    print(feat_extr)
    scores = merge_csvs(os.path.join(output_path,'CD45'),feat_extr, segment='CD45+',ml_models=['SVM'])
    full_scores.append(scores)

cd45_scores = pd.concat(full_scores)
cd45_scores.to_csv(os.path.join(output_path,f"allgenescores_cd45.csv"),index=False)
```

```python
allscores = pd.concat([panck_scores,cd45_scores])
allscores.to_csv(os.path.join(output_path,f"allgenescores.csv"),index=False)
```
