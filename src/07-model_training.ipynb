# 7 - Training & Cross-Validation

This code trains 4 ML models (Logistic Regression, Support Vector Machine, Random Forest and Gradient Boosted Trees) to predict spatial gene expression from foundation model embeddings (Extracted from H&E WSIs).
Five-fold cross validation is used to select optimal hyperparameters per model, per gene.

This step is submitted as a job in a  bash script. The cell below demonstrated the code in the .sh file

```bash
#$ -l tmem=30G
#$ -l gpu=true
#$ -pe gpu 1
#$ -R y
#$ -l h_rt=20:00:00
#$ -j y
#$ -N 'st_cHVG-uni'


host=$(hostname)
echo "host: $host"

date

export LD_LIBRARY_PATH=/usr/lib64:$LD_LIBRARY_PATH
export PATH=/home/hrafique/.local/bin:$PATH
source /share/apps/source_files/python/python-3.11.9.source

input_path="data/LEAP/features"
truth_path="data/LEAP/truthlabels"
output_path="data/LEAP/output"
label_file="hvgs_cd45_raw_v3.csv"
segment="CD45+"
foundation_model="uni"
feature_file="${foundation_model}-${segment}_features.csv"

python3 src/04-predict_sge.py -ip $input_path -tp $truth_path -op $output_path \
                                                     -tf $label_file -ff $feature_file \
                                                     -fm $foundation_model  -s $segment
```
