```python
import numpy as np
import pandas as pd
import os


from utils import load_features_from_csv
```

```python
input_path ='../data/02-input/features'
```

# CD45+

```python
segment='CD45+'
feat_extrs=['uni','virchow','gigapath']
```

```python
segment=''
```

## UNI

```python
feat_file= f'{feat_extrs[0]}-{segment}_features.csv'
df = pd.read_csv(os.path.join(input_path,feat_file))
df = df.set_index('image_name')
print(df.shape)
combined_df = df
```

## Virchow

```python
feat_file= f'{feat_extrs[1]}-{segment}_features.csv'
df = pd.read_csv(os.path.join(input_path,feat_file))
df = df.set_index('image_name')
print(df.shape)
```

```python
combined_df = combined_df.join(df,how='left', lsuffix=f'_{feat_extrs[0]}')
```

## Gigapath

```python
feat_file= f'{feat_extrs[2]}-{segment}_features.csv'
df = pd.read_csv(os.path.join(input_path,feat_file))
df = df.set_index('image_name')
print(df.shape)
```

```python
combined_df = combined_df.join(df,how='left', rsuffix=f'_{feat_extrs[2]}')
```

```python
combined_df.head()
```

## Save

```python
feat_file= f'combined-{segment}_features.csv'
combined_df.to_csv(os.path.join(input_path,feat_file))
```
